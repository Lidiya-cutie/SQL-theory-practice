## Предобработка данных

Мы познакомились с данными. Обнаружили, что наших данных есть пропуски. Выведем информацию о них в процентном соотношении:

```python
df.isnull().mean() * 100
```

У нас отсутствует около 56% информации о средней цене товара, 10% информации о бренде, около 1% информации о потенцальном доходе и упущенной выгоде, а также среднего значения при наличии на складе и менее 1% в категории фото и URL-адрес.


* Удалим столбцы c категориальными данными, в которых более 30 % пропусков
* Заполним оставшиеся ячейки константами

```python
n = df.shape[0]
thresh = n * 0.7 
df = df.dropna(thresh=thresh, axis=1)
values = {
    'Brand': df['Brand'].mode()[0], 
    'Revenue potential': df['Revenue potential'].median(), 
    'Lost profit': df['Lost profit'].median(),
    'Average if in stock': df['Average if in stock'].median(), 
    'URL': df['URL'].mode()[0],
    'Photo': df['Photo'].mode()[0]
    }
df = df.fillna(values)
df.isnull().mean() * 100
```

Теперь в наших данный отсутствуют пропуски. 

На всякий случай воспользуемся функцией удаления дубликатов:

```python
df.drop_duplicates()
```

Теперь переименуем столбцы, чтобы они больше соотвествовали единым требованиям и с ними было проще работать. Для этого заменим пропуски на нижнее подчеркивание и заменим заглавные буквы на строчные:

```python
df.columns = df.columns.str.replace(' ', '_')
df.columns = [x.lower() for x in df.columns]
df.head(3)
```
